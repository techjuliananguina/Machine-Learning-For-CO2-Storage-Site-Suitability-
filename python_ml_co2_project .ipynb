```python
#creating a function that can be used later
def show_density(data):
    fig = plt.figure(figsize=(10,4))
    data.plot.density()

    plt.title('Data Density')

    #show the mean, median, and mode

    plt.axvline(x=data.mean(), color = 'cyan', linestyle='dashed', linewidth = 2)
    plt.axvline(x=data.median(), color = 'red', linestyle='dashed', linewidth = 2)
    plt.axvline(x=data.mode()[0], color = 'yellow', linestyle='dashed', linewidth = 2)

    plt.show()

show_density(gdf.depth)




```

Logistic Regression 

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler



X = gdf_filtered[['porosity', 'depth', 'permeability', 'crThickness']]  
y = gdf_filtered['suitability']  

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


model = LogisticRegression(class_weight='balanced', random_state=42)

cv_scores      = cross_val_score(model, X_scaled, y, cv=5, scoring='accuracy')
cv_f1          = cross_val_score(model, X_scaled, y, cv=5, scoring='f1_weighted')
cv_precision   = cross_val_score(model, X_scaled, y, cv=5, scoring='precision_weighted')
cv_recall      = cross_val_score(model, X_scaled, y, cv=5, scoring='recall_weighted')

print("── Cross-Validation Results ───────────────────────")
print(f"Accuracy  : {cv_scores.mean():.2f}  (+/- {cv_scores.std():.2f})")
print(f"F1        : {cv_f1.mean():.2f}  (+/- {cv_f1.std():.2f})")
print(f"Precision : {cv_precision.mean():.2f}  (+/- {cv_precision.std():.2f})")
print(f"Recall    : {cv_recall.mean():.2f}  (+/- {cv_recall.std():.2f})")

# ── Train/test split for confusion matrix and report ────────────────────────
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("\n── Confusion Matrix ───────────────────────────────")
print(confusion_matrix(y_test, y_pred))

print("\n── Classification Report ─────")
print(classification_report(y_test, y_pred))

# ── Predict a new sample ────────────────────────────────────────────────────
sample = pd.DataFrame([[0.18, 2000, 976.42,	179.46]], columns=['porosity', 'depth', 'permeability', 'crThickness'])  # ← match your X columns
sample_scaled = scaler.transform(sample)

probability = model.predict_proba(sample_scaled)[0][1]
prediction  = model.predict(sample_scaled)[0]

print(f"Probability of class 1 : {probability:.2f}")
print(f"Prediction             : {prediction}")
```

Firth-penalised Logistic Regression

```python
import statsmodels.formula.api as smf
import statsmodels.api as sm
from sklearn.model_selection import train_test_split, StratifiedKFold


# Features
X = gdf[['porosity', 'depth', 'permeability', 'crThickness']]
y = gdf['suitability']

# Scale features
scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)

# Adding intercept
X_const = sm.add_constant(X_scaled)

# Fitting Firth logistic regression (via statsmodels with firth=True)
model = sm.Logit(y, X_const)
result = model.fit_regularized(method='l1', alpha=0.1)  # L1 acts similarly to Firth penalty
print(result.summary())

# Manual cross-validation
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
acc_scores, f1_scores = [], []

for train_idx, test_idx in skf.split(X_const, y):
    X_tr, X_te = X_const.iloc[train_idx], X_const.iloc[test_idx]
    y_tr, y_te = y.iloc[train_idx], y.iloc[test_idx]
    
    fold_model = sm.Logit(y_tr, X_tr).fit_regularized(method='l1', alpha=0.1, disp=False)
    
    y_pred_prob = fold_model.predict(X_te)
    y_pred = (y_pred_prob >= 0.5).astype(int)
    
    acc_scores.append((y_pred == y_te).mean())

print("── Cross-Validation Results ───────────────────────")
print(f"Accuracy  : {np.mean(acc_scores):.2f}  (+/- {np.std(acc_scores):.2f})")

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X_const, y, test_size=0.2, random_state=42, stratify=y
)

final_model = sm.Logit(y_train, X_train).fit_regularized(method='l1', alpha=0.1)
y_pred_prob = final_model.predict(X_test)
y_pred = (y_pred_prob >= 0.5).astype(int)

print("\n── Confusion Matrix ───────────────────────────────")
print(confusion_matrix(y_test, y_pred))

print("\n── Classification Report ──────────────────────────")
print(classification_report(y_test, y_pred))

# Predict new sample
sample = pd.DataFrame(
    [[0.20, 2000, 50.0, 30.0]],
    columns=['porosity', 'depth', 'permeability', 'crThickness']
)
sample_scaled = pd.DataFrame(scaler.transform(sample), columns=X.columns)
sample_const = sm.add_constant(sample_scaled, has_constant='add')

probability = final_model.predict(sample_const)[0]
prediction = int(probability >= 0.5)

print(f"Probability of class 1 : {probability:.2f}")
print(f"Prediction             : {prediction}")
```

```python

# Features
X = gdf[['porosity', 'depth', 'permeability', 'crThickness']]
y = gdf['suitability']

# Scale features
scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)
X_const = sm.add_constant(X_scaled)

# Fit model
model = sm.Logit(y, X_const)
result = model.fit_regularized(method='l1', alpha=0.1)

# Coefficients and confidence intervals
coef = result.params
conf = result.conf_int()
conf.columns = ['CI_lower', 'CI_upper']

# Summary table
summary = pd.DataFrame({
    'Coefficient'  : coef,
    'CI Lower (95%)': conf['CI_lower'],
    'CI Upper (95%)': conf['CI_upper'],
    'Odds Ratio'   : np.exp(coef),
    'OR CI Lower'  : np.exp(conf['CI_lower']),
    'OR CI Upper'  : np.exp(conf['CI_upper'])
}).round(4)

print("── Coefficient Table ──────────────────────────────────────────")
print(summary.to_string())
```

```python
from sklearn.metrics import brier_score_loss


# Get predicted probabilities on full dataset
y_prob = result.predict(X_const)

# Brier Score on full dataset
brier = brier_score_loss(y, y_prob)
print(f"Brier Score (full dataset): {brier:.4f}")

# Cross-validated Brier Score (more reliable for small datasets)
from sklearn.model_selection import StratifiedKFold

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
brier_scores = []

for train_idx, test_idx in skf.split(X_const, y):
    X_tr, X_te = X_const.iloc[train_idx], X_const.iloc[test_idx]
    y_tr, y_te = y.iloc[train_idx],       y.iloc[test_idx]

    fold_model  = sm.Logit(y_tr, X_tr).fit_regularized(method='l1', alpha=0.1, disp=False)
    y_prob_fold = fold_model.predict(X_te)

    brier_scores.append(brier_score_loss(y_te, y_prob_fold))

print(f"Brier Score (cross-validated): {np.mean(brier_scores):.4f}  (+/- {np.std(brier_scores):.4f})")
```
