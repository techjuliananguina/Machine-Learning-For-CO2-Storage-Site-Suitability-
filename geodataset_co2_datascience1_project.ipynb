```python
from geopy.distance import geodesic
```

```python
# a function to compute nearest site

def distance_to_emissions(lat, lon, df):
    #  site coordinates
    coords = list(zip(df["Latitude"], df["Longitude"]))
    
    # distances (in km)
    dists = [geodesic((lat, lon), (site_lat, site_lon)).km for site_lat, site_lon in coords]
    
    # Find nearest
    min_dist = min(dists)
    idx = dists.index(min_dist)   
    
    return min_dist, idx  
```

```python
# usage

#  nearest emitter to a test location (e.g., London)
lat, lon = 51.5074, -0.1278

dist, idx = distance_to_emissions(lat, lon, df)

print(f"Nearest site is at row {idx}, distance = {dist:.2f} km")
print(df.iloc[idx])
```
